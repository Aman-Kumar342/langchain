```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
final_documents=text_splitter.split_documents(docs)
print(final_documents[0])
print(final_documents[1])
```

```python
# HTMLHeaderSplitter

from langchain_text_splitters import HTMLHeaderTextSplitter
html_string="""<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <title>LangChain Demo Page</title>
</head>
<body>

  <h1>LangChain Overview</h1>
  <p>
    LangChain is a framework designed to build applications powered by large language models.
    It helps developers connect LLMs with external data and tools.
  </p>

  <h2>Why LangChain?</h2>
  <p>
    LangChain simplifies prompt management, chaining, memory, and integrations.
  </p>

  <h2>Core Components</h2>

  <h3>Document Loaders</h3>
  <p>
    Document loaders help ingest data from PDFs, HTML, CSVs, and websites.
  </p>

  <h3>Text Splitters</h3>
  <p>
    Text splitters break large documents into smaller, manageable chunks.
  </p>

  <h2>Use Cases</h2>
  <p>
    LangChain is widely used for chatbots, RAG systems, and AI agents.
  </p>

</body>
</html>
"""
header_to_split_on=[("h1","Header 1"),("h2","Header 2"),("h3","Header 3")]

html_splitter=HTMLHeaderTextSplitter(header_to_split_on)
html_header_splits=html_splitter.split_text(html_string)
html_header_splits
```

```python
# JSON SPlitter

%%writefile sample.json
{
  "company": "GenAI Solutions",
  "products": [
    {
      "id": 1,
      "name": "ChatAssist",
      "category": "Chatbot",
      "features": [
        "Natural language understanding",
        "Context-aware responses",
        "Multi-language support"
      ]
    },
    {
      "id": 2,
      "name": "DocSearch",
      "category": "RAG",
      "features": [
        "PDF ingestion",
        "Vector search",
        "Answer generation"
      ]
    }
  ],
  "security": {
    "authentication": "OAuth2",
    "encryption": "AES-256"
  }
}
```

```python
from langchain_community.document_loaders import JSONLoader

loader = JSONLoader(
    file_path="sample.json",
    jq_schema=".products[]",   # split per product
    text_content=False
)

documents = loader.load()

print("Documents loaded:", len(documents))
print(documents[0].page_content)
```
