```python
# logging is a crucial aspect of any application providing a way to track events , error and operational informaion. python built in logging module offers a flexible framework for emitting log message from python programs . in this lesson we will cover the basics of logging including how to configure logging log levels and best practice for using logging in python application
```

```python
import logging
# configure the basics logging srttings
logging.basicConfig(level=logging.DEBUG)

# log messgae
logging.debug("this is a debug messge")
logging.info("this is a info message")
logging.warning("this is a warning message")
logging.error("this is an error message")
logging.critical("this is an critical messgae")
```

```python
# configure the login
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s %(name)s %(levelname)s %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S",
    filename='app.log',
    filemode='w'
)
logging.debug("this is a debug messge")
logging.info("this is a info message")
logging.warning("this is a warning message")
logging.error("this is an error message")
logging.critical("this is an critical messgae")
```

```python
import logging
# logging setting
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s %(name)s %(levelname)s %(message)s',
    datefmt='%Y-%m-%d',
    handlers=[
        logging.FileHandler('app.log'),
        logging.StreamHandler()
    ]
)

logger=logging.getLogger("arithmentic app")

def add(a,b):
  result=a+b
  logger.debug(f"Adding {a}+{b}",result)
  return result
def sub(a,b):
  result=a-b
  logger.debug(f"subtarcting {a}-{b}",result)
  return result
def multiply(a,b):
  result=a*b
  logger.debug(f"multiply {a}*{b}",result)
  return result
def divide(a,b):
  result=a//b
  logger.debug(f"dividing {a}{b}",result)
  return result
```
